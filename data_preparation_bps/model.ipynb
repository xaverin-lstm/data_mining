# import library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# membaca dataset

```python
df = pd.read_csv("data/jumlah_desa.csv")
df.head()
```

```python
df.info()
```

```python
print(df.columns)  # Cek daftar kolom dalam DataFrame
```

# normalisasi kolom

```python
# mengubah nama kolom
df.columns = ["Provinsi", "Lembah", "Lereng_Puncak", "Dataran"]

# isi missing values dengan 0
df.fillna(0, inplace=True)

# mengubah tipe data menjadi integer
df[["Lembah", "Lereng_Puncak", "Dataran"]] = df[["Lembah", "Lereng_Puncak", "Dataran"]].astype(int)

# menambahkan kolom Total_Desa
df["Total_Desa"] = df["Lembah"] + df["Lereng_Puncak"] + df["Dataran"]

df.head()
```

# melakukan visualisasi

```python
# diurutkan berdasarkan jumlah desa terbanyak

plt.figure(figsize=(12,6))
sns.barplot(x='Total_Desa', y='Provinsi', data=df.sort_values('Total_Desa', ascending=False), palette='coolwarm')
plt.xlabel("Total Desa")
plt.ylabel("Provinsi")
plt.title("Jumlah Desa per Provinsi")
plt.show()
```

```python
# ini kalo mau urut sesuai dataframe yg dari csv (ga dimasukin ke laprak)

plt.figure(figsize=(12, 8))
sns.barplot(
    x="Total_Desa", 
    y="Provinsi", 
    data=df, 
    palette="coolwarm"
)
plt.xlabel("Total Desa")
plt.ylabel("Provinsi")
plt.title("Jumlah Desa per Provinsi")
plt.show()
```

# mengapus data "Indonesia"

```python
df = df[df["Provinsi"] != "Indonesia"]

df.head()
```

# melakukan visualisasi data

```python
plt.figure(figsize=(12,6))
sns.barplot(x='Total_Desa', y='Provinsi', data=df.sort_values('Total_Desa', ascending=False), palette='coolwarm')
plt.xlabel("Total Desa")
plt.ylabel("Provinsi")
plt.title("Jumlah Desa per Provinsi")
plt.show()
```
