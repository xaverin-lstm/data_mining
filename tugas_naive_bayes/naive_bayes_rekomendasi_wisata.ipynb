# Naive Bayes untuk Rekomendasi Tempat Wisata di Yogyakarta

```python
# Import library
import pandas as pd

# Load data tempat wisata
df = pd.read_csv("data/package_tourism.csv")
df.head()
```

## 💡 Beri label secara manual pada data berikut

```python
# Contoh: Tambahkan kolom 'Label' dengan nilai 'recommended' atau 'not_recommended'
df['Label'] = [
    'recommended', 'recommended', 'recommended', 'recommended', 'recommended',
    'not_recommended', 'not_recommended', 'not_recommended', 'not_recommended', 'not_recommended',
    'recommended', 'not_recommended', 'recommended', 'not_recommended', 'recommended',
    'not_recommended', 'not_recommended', 'not_recommended', 'recommended', 'recommended',
    'recommended', 'not_recommended', 'not_recommended', 'not_recommended', 'recommended',
    'not_recommended', 'not_recommended', 'recommended', 'recommended', 'not_recommended'
]  # Ini contoh, silakan ubah sesuai preferensi

df.head()
```

## 🔍 Training model Naive Bayes berdasarkan label

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# Vektorisasi nama tempat wisata
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['Place'])

# Label
y = df['Label']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Model Naive Bayes
model = MultinomialNB()
model.fit(X_train, y_train)

# Evaluasi
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
```
