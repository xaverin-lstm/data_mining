```python
# Import library

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, accuracy_score
```

```python
# Load dataset

df = pd.read_csv('data/tempat_wisata_yogyakarta.csv')
df.head()
```

```python
# Membuat kolom label

df['recommended'] = ((df['vote_average'] >= 4.5) & (df['vote_count'] >= 10000)).astype(int)
df[['vote_average', 'vote_count', 'recommended']].head()
```

```python
# Pilih fitur dan target

features = ['vote_average', 'vote_count', 'htm_weekday', 'htm_weekend']
X = df[features].fillna(0)  # isi NaN dengan 0
y = df['recommended']
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
print('Akurasi:', accuracy_score(y_test, y_pred))
print('\nLaporan Klasifikasi:\n', classification_report(y_test, y_pred, target_names=['Tidak Rekomendasi', 'Rekomendasi']))
```

```python
from imblearn.over_sampling import RandomOverSampler

ros = RandomOverSampler(random_state=42)
X_resampled, y_resampled = ros.fit_resample(X, y)

# Training ulang
model = GaussianNB()
model.fit(X_resampled, y_resampled)
y_pred = model.predict(X_test)

print('Akurasi:', accuracy_score(y_test, y_pred))
print('\nLaporan Klasifikasi:\n', classification_report(y_test, y_pred, target_names=['Tidak Rekomendasi', 'Rekomendasi']))
```
