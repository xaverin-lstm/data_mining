```python
import tweepy

bearer_token = 'BEARER TOKEN'

client = tweepy.Client(bearer_token)

def get_tweets(query, count=100):
    tweet_data = []
    tweets = client.search_recent_tweets(query=query, max_results=count, tweet_fields=["created_at", "text"])
    for tweet in tweets.data:
        tweet_data.append(tweet.text)
    return tweet_data

tweets = get_tweets('AI', count=100)
for tweet in tweets:
    print(tweet)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
nltk.download('stopwords')

stop_words = set(stopwords.words('indonesian'))
stemmer = PorterStemmer()

def preprocess(text):
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    
    text = text.lower()
    
    words = text.split()
    words = [word for word in words if word not in stop_words]
    
    words = [stemmer.stem(word) for word in words]
    
    return ' '.join(words)

processed_tweets = [preprocess(tweet) for tweet in tweets]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=1000)
X = vectorizer.fit_transform(processed_tweets)
```

```python
from sklearn.cluster import KMeans

# Tentukan jumlah cluster (misalnya, 5 cluster)
num_clusters = 5
kmeans = KMeans(n_clusters=num_clusters, random_state=42)

# Latih model clustering
kmeans.fit(X)

# Prediksi cluster untuk setiap tweet
clusters = kmeans.predict(X)

# Tampilkan hasil cluster
for i in range(num_clusters):
    print(f"\nCluster {i + 1}:")
    for index, tweet in enumerate(tweets):
        if clusters[index] == i:
            print(f" - {tweet}")
```

```python
from sklearn.metrics import silhouette_score

# Hitung Silhouette Score
score = silhouette_score(X, clusters)
print(f"Silhouette Score: {score}")
```

```python
import pandas as pd

# Simpan hasil clustering ke dalam DataFrame
df = pd.DataFrame({'tweet': tweets, 'cluster': clusters})

# Simpan ke file CSV
df.to_csv('clustering_results.csv', index=False)
```
